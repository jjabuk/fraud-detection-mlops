## 4. Why Unpinned Bins Fail

Refitting the binning on the current sample defines every bucket to hold a tenth of *it*. The sample then compares as unshifted against itself, no matter how far it moved.

````python
refit = Reference.fit(test, ["D15"])  # edges from the current sample instead

print(f"D15 PSI, edges pinned to train : {ref.psi(test).set_index('column').loc['D15', 'psi']:.4f}")
print(f"D15 PSI, edges refit on test   : {refit.psi(test).set_index('column').loc['D15', 'psi']:.4f}")
````

Same data, opposite conclusion. The largest shift in the dataset reads as zero.

This is the reason `Reference` is a serializable object rather than a function argument: `ref.to_dict()` is the artefact you commit, and it is what makes a PSI computed months apart mean anything.

## Architecture

| Caller | Reference | Current |
| --- | --- | --- |
| Feature admission | training period | held-out later period |
| Drift monitoring | pinned training baseline | serving window |
| Diagnostics | training period | later period — read the importances, not the AUC, to find columns encoding *when* a row arrived rather than what it is |

# ieee-cis-fraud-detection distribution shift

Two questions about drift that get treated as one — and what they actually say about the
IEEE-CIS train/test split.

*PSI* asks whether a column's marginal distribution moved. *Adversarial validation* asks
whether a model can tell the two samples apart at all. Same input, different question, and
production monitors routinely report one while reasoning about the other.

```python
from distribution_shift import Reference, adversarial_auc, adversarial_per_feature

ref = Reference.fit(train, columns)      # pins the bucket edges
ref.psi(current)                          # per-column PSI against those edges
adversarial_auc(train, current, columns)  # out-of-fold AUC + what drives it
```

## What came out

377 columns — `C1–C14`, `D1–D15`, `M1–M9`, `V1–V339` — reference fitted on
`train_transaction.csv` (590,540 rows), compared against `test_transaction.csv`
(506,691 rows). `TransactionDT` is deliberately excluded: including the timestamp makes
any discriminator trivially perfect and tells you nothing.

| PSI band | Columns |
| --- | ---: |
| > 0.25 ("significant" by convention) | 19 |
| 0.10 – 0.25 | 63 |
| ≤ 0.10 | 218 |
| degenerate (constant in the reference) | 77 |

Adversarial validation over all 377 at once: **out-of-fold AUC 0.838**.

### The largest shift in the dataset is missingness, not values

Every one of the top PSI columns is driven by the same thing — `D15`, then `V77`–`V93` —
and it is not their values:

| Column | PSI | Null share, train | Null share, test |
| --- | ---: | ---: | ---: |
| `D15` | 0.339 | 15.1% | **2.4%** |
| `V85` | 0.256 | 15.1% | **2.4%** |
| `V81` | 0.256 | 15.1% | **2.4%** |
| `V78` | 0.254 | 15.1% | **2.4%** |

A block of columns went from 15% missing to 2% missing. Something upstream started
populating them. Nothing about the values needed to move for this to be the dominant
signal in the whole comparison.

This is why the library treats **missingness as a bucket rather than as rows to drop**. A
PSI implementation that filters nulls before bucketing — which is the common shortcut —
would miss the single largest distribution shift in this dataset entirely, and report the
same columns as stable.

### The two methods mostly agree, and where they don't is asymmetric

Going in, the expectation was that PSI and adversarial validation would disagree. Measured
across the 300 evaluable columns, **Spearman correlation is 0.857** — they mostly agree,
and the hypothesis was wrong.

What survives is one-directional:

- **12 columns** score PSI > 0.25 while a discriminator built on them alone reaches
  under 0.60 AUC.
- **0 columns** go the other way.

So on this data PSI is the more trigger-happy of the two: it flags marginal movement that
carries no usable information about which sample a row came from. If your monitor pages
someone on PSI alone, those twelve columns are the shape of the false alarm.

### Discriminability is combinatorial, and that breaks per-column monitoring

No single column dates a row well. The best is `D15` at **0.656** out-of-fold, and the
per-column AUCs fall away quickly from there. Yet all 377 together reach **0.838**.

The separation lives in the *combination*, not in any column. A drift monitor that
iterates columns one at a time and alerts on thresholds will not see it — not because the
thresholds are wrong, but because the signal is not in any of the marginals it is looking
at.

## Pinned edges, and why the `Reference` is the artefact

`Reference.fit` stores the bucket edges and the reference shares inside them.
`Reference.to_dict()` serializes both; commit that file.

That is not a convenience. Recomputing quantile edges on the current sample defines every
bucket to hold a tenth of it, so **the sample compares as unshifted against itself no
matter how far it moved**. There is a test that asserts exactly this: the same shifted
frame gives PSI > 0.25 against pinned edges and < 1e-9 against refitted ones. Same data,
opposite conclusion.

## Out-of-fold, not in-sample

`adversarial_auc` cross-fits. This is not fussiness: an in-sample discriminator with a few
hundred trees reports **AUC ≈ 0.63 on two samples drawn from the same distribution** —
measured, and pinned by a test. An adversarial check scored in-sample cannot tell "nothing
moved" from "something did", which makes it a false-alarm generator wearing a metric's
name.

## Cost

PSI over 377 columns is seconds — it is counting. The adversarial side is model fitting:
on a 100k/100k subsample, one 377-column discriminator takes ~95 s, and
`adversarial_per_feature(..., n_jobs=-1)` over all 377 takes ~340 s on eight cores.

Sample the rows for the adversarial half. Doubling the rows does not change which columns
separate the samples, and the fit cost is linear in them.

## Where this belongs in a pipeline

The same implementation answers three questions at three points in a model's life, which
is the reason to write it once:

| Caller | Reference | Current |
| --- | --- | --- |
| Feature admission | training period | held-out later period |
| Drift monitoring | pinned training baseline | serving window |
| Diagnostics | training period | later period — read the *importances*, not the AUC, to find columns that encode when a row arrived rather than what it is |

Run it as an audit that produces a committed artefact, not as a gate inside every training
run: the answer is a property of the data, and re-deriving it per run buys nothing.
